```python
# ============================================
# STEP 4.1: CREATE WORKING COPIES
# ============================================

school = df_school.copy()
attendance = df_attendance.copy()
mdm = df_mdm.copy()
scores = df_scores.copy()
infra = df_infra.copy()

print("Working copies created.")
```

```python
# ============================================
# STEP 4.2: STANDARDIZE COLUMN NAMES
# ============================================

def clean_column_names(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )
    return df


school = clean_column_names(school)
attendance = clean_column_names(attendance)
mdm = clean_column_names(mdm)
scores = clean_column_names(scores)
infra = clean_column_names(infra)

print("Column names standardized.")
```

```python
# ============================================
# STEP 4.3: STANDARDIZE SCHOOL IDs
# ============================================

def standardize_school_id(series):
    return (
        series
        .astype("string")
        .str.strip()
        .str.upper()
        .str.replace("-", "", regex=False)
        .str.replace("_", "", regex=False)
        .str.replace(" ", "", regex=False)
    )


for df in [school, attendance, mdm, scores, infra]:
    df["school_id"] = standardize_school_id(df["school_id"])


print("School IDs standardized.")

print("\nExample standardized IDs:")
print(school["school_id"].head(10).tolist())
```

```python
# ============================================
# STEP 10: STANDARDIZE ATTENDANCE DATES
# ============================================

import re
from datetime import datetime

def parse_attendance_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # ISO format: YYYY-MM-DD / YYYY/MM/DD / YYYY.MM.DD
    if re.match(r"^\d{4}[-/.]\d{1,2}[-/.]\d{1,2}$", value):
        return pd.to_datetime(value, format="mixed", errors="coerce")

    # Text month format: 11-Sep-2025
    if re.match(r"^\d{1,2}-[A-Za-z]{3}-\d{4}$", value):
        return pd.to_datetime(value, format="%d-%b-%Y", errors="coerce")

    # Numeric date formats
    parts = re.split(r"[-/.]", value)

    if len(parts) == 3:
        a, b, c = parts

        try:
            a, b, c = int(a), int(b), int(c)
        except ValueError:
            return pd.NaT

        # If first component is 4-digit year
        if len(str(parts[0])) == 4:
            return pd.to_datetime(
                f"{a}-{b:02d}-{c:02d}",
                format="%Y-%m-%d",
                errors="coerce"
            )

        # If first component > 12, it must be DD-MM-YYYY
        if a > 12:
            return pd.to_datetime(
                f"{c}-{b:02d}-{a:02d}",
                format="%Y-%m-%d",
                errors="coerce"
            )

        # If second component > 12, it must be MM-DD-YYYY
        if b > 12:
            return pd.to_datetime(
                f"{c}-{a:02d}-{b:02d}",
                format="%Y-%m-%d",
                errors="coerce"
            )

        # Ambiguous case: use month-first for consistency
        return pd.to_datetime(
            f"{c}-{a:02d}-{b:02d}",
            format="%Y-%m-%d",
            errors="coerce"
        )

    return pd.NaT


attendance["date_clean"] = attendance["date"].apply(parse_attendance_date)

print("========== DATE CLEANING AUDIT ==========")
print("Original rows:", len(attendance))
print("Valid dates:", attendance["date_clean"].notna().sum())
print("Invalid dates:", attendance["date_clean"].isna().sum())

print("\nDate range:")
print("Minimum:", attendance["date_clean"].min())
print("Maximum:", attendance["date_clean"].max())

print("\nMissing date values:", attendance["date_clean"].isna().sum())

print("\nSample standardized dates:")
display(
    attendance[["date", "date_clean"]]
    .drop_duplicates()
    .head(20)
)
```

```python
# ============================================
# STEP 11: SCHOOL ID NORMALIZATION AUDIT
# ============================================

import re

def normalize_school_id(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().upper()

    # Extract numeric part
    numbers = re.findall(r"\d+", value)

    if len(numbers) == 1:
        return f"SCH{int(numbers[0]):04d}"

    return pd.NA


attendance["school_id_clean"] = (
    attendance["school_id"]
    .apply(normalize_school_id)
)

print("========== SCHOOL ID NORMALIZATION AUDIT ==========")

print("Original unique school IDs:",
      attendance["school_id"].nunique())

print("Normalized unique school IDs:",
      attendance["school_id_clean"].nunique())

print("\nInvalid / unmapped IDs:",
      attendance["school_id_clean"].isna().sum())

print("\n========== RAW → CLEANED SAMPLE ==========")

display(
    attendance[["school_id", "school_id_clean"]]
    .drop_duplicates()
    .head(30)
)

print("\n========== NORMALIZED ID FREQUENCY ==========")

print(
    attendance["school_id_clean"]
    .value_counts()
    .head(20)
)
```

```python
# ============================================
# STEP 16: REMOVE EXACT DUPLICATE ROWS
# ============================================

rows_before = len(attendance)

duplicate_count = attendance.duplicated().sum()

attendance = attendance.drop_duplicates().copy()

rows_after = len(attendance)

print("========== DUPLICATE CLEANING ==========")
print("Rows before:", rows_before)
print("Exact duplicate rows removed:", duplicate_count)
print("Rows after:", rows_after)

print("\nRemaining exact duplicates:",
      attendance.duplicated().sum())

print("\nRemaining impossible attendance records:",
      attendance["impossible_attendance_flag"].sum())

print("\nRemaining missing record IDs:",
      attendance["record_id"].isna().sum())
```

```python
# ============================================
# STEP 17: RECORD ID AUDIT
# ============================================

print("========== RECORD ID AUDIT ==========")

print("Total rows:", len(attendance))

print("Missing record IDs:",
      attendance["record_id"].isna().sum())

print("Non-missing record IDs:",
      attendance["record_id"].notna().sum())

print("Unique non-missing record IDs:",
      attendance["record_id"].dropna().nunique())

print("Duplicate non-missing record IDs:",
      attendance.loc[
          attendance["record_id"].notna(),
          "record_id"
      ].duplicated().sum())

# Create a flag rather than inventing IDs
attendance["record_id_missing_flag"] = attendance["record_id"].isna()

print("\n========== DUPLICATE RECORD IDs ==========")

duplicate_record_ids = (
    attendance.loc[
        attendance["record_id"].notna(),
        "record_id"
    ]
    .value_counts()
)

duplicate_record_ids = duplicate_record_ids[
    duplicate_record_ids > 1
]

print("Number of duplicated record IDs:",
      len(duplicate_record_ids))

if len(duplicate_record_ids) > 0:
    print("\nSample duplicated IDs:")
    print(duplicate_record_ids.head(20))
```

```python
# ============================================
# STEP 18: FLAG PROXY ATTENDANCE ANOMALIES
# ============================================

attendance["is_sunday"] = (
    attendance["date_clean"].dt.dayofweek == 6
)

attendance["proxy_attendance_flag"] = (
    attendance["is_sunday"]
    & (attendance["attendance_rate"] == 100)
)

print("========== PROXY ATTENDANCE AUDIT ==========")

print("Sunday records:",
      attendance["is_sunday"].sum())

print("Sunday 100% attendance records:",
      attendance["proxy_attendance_flag"].sum())

print(
    "Proxy attendance rate:",
    round(attendance["proxy_attendance_flag"].mean() * 100, 2),
    "%"
)
```

```python
# ============================================
# STEP 19: FINAL ATTENDANCE VALIDATION
# ============================================

print("========== FINAL ATTENDANCE VALIDATION ==========")

print("Rows:", len(attendance))
print("Columns:", len(attendance.columns))

print("\nMissing values:")
print(attendance.isna().sum())

print("\nExact duplicate rows:")
print(attendance.duplicated().sum())

print("\nImpossible attendance records:")
print(attendance["impossible_attendance_flag"].sum())

print("\nProxy attendance records:")
print(attendance["proxy_attendance_flag"].sum())

print("\nValid attendance rates:")
print(attendance["attendance_rate"].notna().sum())

print("\nAttendance rate range:")
print(
    round(attendance["attendance_rate"].min(), 2),
    "to",
    round(attendance["attendance_rate"].max(), 2)
)

print("\nSchool IDs not in School Master:")
print(
    (~attendance["school_id_clean"].isin(df_school["school_id"]))
    .sum()
)
```

```python
# ============================================
# STEP 20: FINAL CLEANED ATTENDANCE DATASET
# ============================================

attendance_cleaned = attendance[
    [
        "record_id",
        "date_clean",
        "school_id_clean",
        "grade_clean",
        "total_students",
        "present_students",
        "teacher_present_clean",
        "marked_by_clean",
        "attendance_rate",
        "impossible_attendance_flag",
        "proxy_attendance_flag",
        "record_id_missing_flag"
    ]
].copy()

# Rename cleaned columns
attendance_cleaned = attendance_cleaned.rename(columns={
    "date_clean": "date",
    "school_id_clean": "school_id",
    "grade_clean": "grade",
    "teacher_present_clean": "teacher_present",
    "marked_by_clean": "marked_by"
})

print("========== FINAL CLEANED ATTENDANCE ==========")
print("Rows:", len(attendance_cleaned))
print("Columns:", len(attendance_cleaned.columns))

print("\nMissing values:")
print(attendance_cleaned.isna().sum())

print("\nDuplicate rows:",
      attendance_cleaned.duplicated().sum())

print("\nImpossible attendance:",
      attendance_cleaned["impossible_attendance_flag"].sum())

print("\nProxy attendance:",
      attendance_cleaned["proxy_attendance_flag"].sum())
```
